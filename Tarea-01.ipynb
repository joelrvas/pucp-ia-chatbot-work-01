```python
import datasets
import json
```

```python
# Descargar el dataset permitiendo la ejecución de código remoto
banking77 = datasets.load_dataset('PolyAI/banking77', trust_remote_code=True)

# Guardar el dataset en archivos JSON
banking77['train'].to_json('data/banking77_train.json')
banking77['test'].to_json('data/banking77_test.json')
```

```python
# Cargar el archivo JSON del dataset de entrenamiento
with open('data/banking77_train.json', 'r') as f:
    train_data = [json.loads(line) for line in f]


# Crear el archivo nlu.yml
with open('data/nlu.yml', 'w') as f:
    f.write('version: "3.0"\n\n')
    f.write('nlu:\n')
    for example in train_data:
        intent = example['label']
        text = example['text'].replace('\n', '').replace('\\n', '').replace('b\'', '').replace('\'', '')
        f.write(f'- intent: {intent}\n  examples: |\n    - {text}\n')

# Cargar el archivo JSON del dataset de prueba
with open('data/banking77_test.json', 'r') as f:
    test_data = json.load(f)

# Crear el archivo test_nlu.yml
with open('data/test_nlu.yml', 'w') as f:
    f.write('version: "3.0"\n\n')
    f.write('nlu:\n')
    for example in test_data:
        intent = example['label']
        text = example['text'].replace('\n', '').replace('\\n', '').replace('b\'', '').replace('\'', '')
        f.write(f'- intent: {intent}\n  examples: |\n    - {text}\n')

# Verificar y corregir los datos generados
with open('data/nlu.yml', 'r') as f:
    content = f.read()

# Corregir la estructura del archivo
corrected_content = content.replace('b\'', '').replace('\'', '')

with open('data/nlu.yml', 'w') as f:
    f.write(corrected_content)

# Verificar y corregir los datos de prueba
with open('data/test_nlu.yml', 'r') as f:
    content = f.read()

corrected_content = content.replace('b\'', '').replace('\'', '')

with open('data/test_nlu.yml', 'w') as f:
    f.write(corrected_content)
```
